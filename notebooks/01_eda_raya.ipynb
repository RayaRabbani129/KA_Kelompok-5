```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('../dataset/heart.csv')
df.head()
```

```python
print("Jumlah Baris dan Kolom:")
print(df.shape)

print("\nJumlah Baris:")
print(df.shape[0])

print("\nJumlah Kolom:")
print(df.shape[1])
```

```python
print("Daftar Kolom:")

for col in df.columns:
    print(col)
```

```python
df.info()
```

```python
print("Jumlah Missing Value:")
print(df.isnull().sum())
```

```python
duplicate = df.duplicated().sum()

print("Jumlah Data Duplikat:")
print(duplicate)
```

```python
plt.figure(figsize=(6,4))

df['target'].value_counts().plot(kind='bar')

plt.title('Distribusi Target')
plt.xlabel('Target')
plt.ylabel('Jumlah Data')

plt.show()

# 0 = Tidak memiliki penyakit jantung
# 1 = Memiliki penyakit jantung
```

```python
df.hist(figsize=(15,10))

plt.suptitle('Distribusi Seluruh Fitur')

plt.show()
```

```python
corr = df.corr()

corr
```

```python
plt.figure(figsize=(10,8))

plt.imshow(corr)

plt.colorbar()

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title("Heatmap Korelasi")

plt.show()
```

```python
target_corr = corr['target'].sort_values(
    ascending=False
)

target_corr
```

```python
plt.figure(figsize=(8,5))

target_corr.drop('target').plot(
    kind='bar'
)

plt.title('Korelasi Fitur terhadap Target')
plt.ylabel('Nilai Korelasi')

plt.show()
```

```python
df.groupby('target')['age'].mean()
```

```python
df.groupby('target')['age'].mean().plot(
    kind='bar'
)

plt.title('Rata-rata Umur Berdasarkan Target')
plt.ylabel('Usia')

plt.show()
```

```python
df.groupby('target')['chol'].mean()
```

```python
df.groupby('target')['chol'].mean().plot(
    kind='bar'
)

plt.title('Rata-rata Kolesterol Berdasarkan Target')
plt.ylabel('Kolesterol')

plt.show()
```
